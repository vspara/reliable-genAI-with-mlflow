```python
import mlflow
import openai
```

```python
target_text = """
Ghent, Brussesl, Antwerp, Liege, Bruges, Leuven, Namur, Mons, Charleroi, Hasselt
"""
```

```python
# Load the prompt
prompt = mlflow.genai.load_prompt("prompts:/best-belgian-cities/2")

# Use the prompt with an LLM
client = openai.OpenAI()
response = client.chat.completions.create(
    messages=[
        {
            "role": "user",
            "content": prompt.format(cities=target_text),
        }
    ],
    model="gpt-4o-mini",
)
```
