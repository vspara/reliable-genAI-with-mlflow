```python
import mlflow
import openai
```

```python
# Specify the tracking URI for the MLflow server.
mlflow.set_tracking_uri("http://127.0.0.1:5000")
```

```python
input_text = """
Fabrizio Miccoli, Elon Musk, Dario Amodei, Sam Altman, Cristiano Ronaldo, Lionel Messi
"""
```

```python
# Load the prompt
prompt = mlflow.genai.load_prompt("prompts:/best-player/1")

# Use the prompt with an LLM
client = openai.OpenAI()
response = client.responses.create(
    model="gpt-6-luna",
    input=prompt.format(players=input_text),
)
```
